````python
import jax.numpy as jnp
from diffPLOG2TROE.parametrization import FallOff, CollisionEfficiency, MixtureRule

default_collider = FallOff(
    name="H+O2(+M)=HO2(+M)",
    hpl_parameters={"A": 4.650E+12, "n": 0.44, "Ea": 0.0},
    lpl_parameters={"A": 1.737e19, "n": -1.230, "Ea": 0.0},
    falloff_parameters={"A": 0.67, "T3": 1e-30, "T2": 1e30, "T1": 1e30},
    falloff_type="troe",
    efficiencies=[
        CollisionEfficiency(name="H2", value=1.30),
        CollisionEfficiency(name="H2O", value=10.00),
    ],
)
ar_specific_rate = FallOff(
    name="H+O2(+AR)=HO2(+AR)",
    hpl_parameters={"A": 4.650E+12, "n": 0.44, "Ea": 0.0},
    lpl_parameters={"A": 6.81e18, "n": -1.2, "Ea": 0.0},
    falloff_parameters={"A": 0.7, "T3": 1e-30, "T2": 1e30, "T1": 1e30},
    falloff_type="troe",
)
he_specific_rate = FallOff(
    name="H+O2(+HE)=HO2(+HE)",
    hpl_parameters={"A": 4.650E+12, "n": 0.44, "Ea": 0.0},
    lpl_parameters={"A": 9.192, "n": -1.2, "Ea": 0.0},
    falloff_parameters={"A": 0.59, "T3": 1e-30, "T2": 1e30, "T1": 1e30},
    falloff_type="troe",
)

extended_falloff_reaction = MixtureRule(
    default_rate_constant = default_collider,
    explicit_rate_constants = {
        "Ar": ar_specific_rate,
        "He": he_specific_rate
    },
    efficiencies = [
        CollisionEfficiency(name="H2", value=1.30),
        CollisionEfficiency(name="H2O", value=10.00)
    ],
    linear = True,
    reduced_pressure = False,
    name = "H+O2(+M)=HO2(+M)",
)

T_range = jnp.linspace(300, 3000, 10)
ciao=[]
for t in T_range:
    ciao.append(extended_falloff_reaction.rate_constant(t, 1, {"Ar": 1})/1000)
````

````python
import matplotlib.pyplot as plt

cantera=[2.7259679874e+08,6.0742605016e+07,2.5087602864e+07,1.3378300719e+07,8.2105097806e+06,5.5082163591e+06,3.9297872883e+06,2.9329026440e+06,2.2655948312e+06,1.7983309372e+06]
error = []
for i in range(len(cantera)):
    error.append((ciao[i] - cantera[i]) / ciao[i])

fig, axs = plt.subplots(1, 2)

axs[0].plot(T_range, cantera, "ro")
axs[0].plot(T_range, ciao, "bo")

axs[1].plot(T_range, error)
plt.show()
````
