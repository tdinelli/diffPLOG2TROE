```python
# --------------------------------------------------------
# External packages
import matplotlib.pyplot as plt
import numpy as np
import jax.numpy as jnp
import equinox as eqx

# --------------------------------------------------------
# KiRATE modules
from KiRATE.kinetics import FallOff

troe_reaction = FallOff(
    name="2CH3(+M)=C2H6(+M)",
    hpl_parameters={"A": 6.770E+16, "n": -1.180, "Ea": 654.00},
    lpl_parameters={"A": 3.400E+41, "n": -7.030, "Ea": 2762.00},
    falloff_parameters={"A": 0.619, "T3": 73.20, "T1": 1180, "T2": 9999.00},
    falloff_type="troe",
    efficiencies={"H2": 2, "CO": 2, "CO2": 3, "H2O": 5},
)
```

```python
wrapper_function = lambda m, t, p, c: m.rate_constant(t, p, c)

grad = eqx.filter_grad(wrapper_function)(troe_reaction, 500, 1, None)
print(grad._falloff_parameters["T3"])
```
