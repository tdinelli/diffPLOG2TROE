# Mixture Rules for Multi-Collider Systems

This notebook demonstrates the use of the `MixtureRule` class for modeling reactions with collider-specific rate constants, including Linear Mixture Rules (LMR-R and LMR-P).

````python
from KiRATE.kinetics import Arrhenius, Plog, MixtureRule, FallOff
````

````python
default = Plog(
    name="H+O2(+Ar)=HO2(+Ar)",
    parameters={
        1.316e-02: {"A": 9.39968e14, "n": -2.14348e00, "Ea": 7.72730e01},
        1.316e-01: {"A": 1.07254e16, "n": -2.15999e00, "Ea": 1.30239e02},
        3.947e-01: {"A": 3.17830e16, "n": -2.15813e00, "Ea": 1.66994e02},
        1.000e00: {"A": 7.72584e16, "n": -2.15195e00, "Ea": 2.13473e02},
        3.000e00: {"A": 2.11688e17, "n": -2.14062e00, "Ea": 2.79031e02},
        1.000e01: {"A": 6.53093e17, "n": -2.13213e00, "Ea": 3.87493e02},
        3.000e01: {"A": 1.49784e18, "n": -2.10026e00, "Ea": 4.87579e02},
        1.000e02: {"A": 3.82218e18, "n": -2.07057e00, "Ea": 6.65984e02},
    },
)
helium = Arrhenius(name="H+O2(+HE)=HO2(+HE)", parameters={"A": 3.37601e-01, "n": 1.82568e-01, "Ea": 3.62408e01})
nitrogen = Arrhenius(name="H+O2(+N2)=HO2(+N2)", parameters={"A": 1.24932e02, "n": -5.93263e-01, "Ea": 5.40921e02})
hydrogen = Arrhenius(name="H+O2(+H2)=HO2(+H2)", parameters={"A": 3.13717e04, "n": -1.25419e00, "Ea": 1.12924e03})
carbon_dioxyde = Arrhenius(
    name="H+O2(+CO2)=HO2(+CO2)", parameters={"A": 1.62413e08, "n": -2.27622e00, "Ea": 1.97023e03}
)
ammonia = Arrhenius(name="H+O2(+NH3)=HO2(+NH3)", parameters={"A": 4.97750e00, "n": 1.64855e-01, "Ea": -2.80351e02})
water = Arrhenius(name="H+O2(+H2O)=HO2(+H2O)", parameters={"A": 3.69146e01, "n": -7.12902e-02, "Ea": 3.19087e01})

singal_rate = MixtureRule(
    name="H+O2(+M)=HO2(+M)",
    linear=True,
    reduced_pressure=True,
    default_rate_constant=default,
    efficiencies={"He": helium, "N2": nitrogen, "H2": hydrogen, "CO2": carbon_dioxyde, "NH3": ammonia, "H2O": water},
)

singal_rate._efficiencies
````

````python
composition = {"N2": 0.5, "Ar": 0.5}
print("3.291710e+11")
print(f"{singal_rate.rate_constant(300, 1, composition):12e}")

print("4.047259e+09")
print(f"{singal_rate.rate_constant(300, 0.001, composition):12e}")

print("3.206421e+11")
print(f"{singal_rate.rate_constant(300, 1, {'He': 0.4, 'N2': 0.5, 'Ar': 0.1}):12e}")
````

````python
# Now compute the Linear Burke rate constant
T = 300
P = 1

print("=" * 60)
print("Computing Linear Burke Rate Constant")
print("=" * 60)

# Step 1: Compute eps for each collider
collider_eps = {}
for species, eff_obj in efficiencies.items():
    if species in composition:
        eps_i = eff_obj.rate_constant(T)  # Efficiency is temperature-dependent
        collider_eps[species] = eps_i
        print(f"eps_{species} = {eps_i:.6f}")

# Step 2: Compute sigmaX_M (mole fraction of species not explicitly listed)
sigmaX_M = 1.0
for species in collider_eps.keys():
    if species in composition:
        sigmaX_M -= composition[species]

print(f"\nsigmaX_M (X_Ar) = {sigmaX_M:.6f}")

# Step 3: Compute eps_mix
eps_mix = sigmaX_M * 1.0  # eps_M = 1 always (for Ar and other unspecified species)
for species, eps_i in collider_eps.items():
    if species in composition:
        eps_mix += composition[species] * eps_i

print(f"eps_mix = {eps_mix:.6f}")

# Step 4: Compute contributions from each collider
k_total = 0.0

print("\n" + "=" * 60)
print("Individual Contributions:")
print("=" * 60)

# N2 contribution (has efficiency only, uses M's rate)
if "N2" in composition:
    eps_N2 = collider_eps["N2"]
    X_N2 = composition["N2"]

    # N2 uses M's rate with effective pressure
    P_eff_N2 = P * eps_mix / 1.0  # eps2 = eps_M = 1
    k_M_N2 = default.rate_constant(T, P_eff_N2)

    contrib_N2 = k_M_N2 * eps_N2 * X_N2 / eps_mix
    k_total += contrib_N2

    print(f"\nN2:")
    print(f"  P_eff = {P_eff_N2:.6f} atm")
    print(f"  k_M(P_eff) = {k_M_N2:.4e}")
    print(f"  contribution = {contrib_N2:.4e}")

# Ar (M) contribution
X_Ar = sigmaX_M
P_eff_M = P * eps_mix / 1.0  # eps_M = 1
k_M = default.rate_constant(T, P_eff_M)

contrib_Ar = k_M * 1.0 * X_Ar / eps_mix  # eps_M = 1
k_total += contrib_Ar

print(f"\nAr (M):")
print(f"  P_eff = {P_eff_M:.6f} atm")
print(f"  k_M(P_eff) = {k_M:.4e}")
print(f"  contribution = {contrib_Ar:.4e}")

ciao = singal_rate.rate_constant(T, P, composition)
# Final result
print("\n" + "=" * 60)
print(f"{329171038824.9761:.15e}")
print(f"{ciao:15e}")
print(f"{k_total:.15e}")
print("=" * 60)
````

````python
import math

# Given data
T = 300  # K
P = 1.0  # atm
R = 1.987  # cal/(mol·K)
composition = {"N2": 0.5, "Ar": 0.5}

print("=" * 60)
print("STEP-BY-STEP CALCULATION")
print("=" * 60)

# Step 1: Compute eps_N2
n2_params = {"A": 1.24932e02, "n": -5.93263e-01, "Ea": 5.40921e02}
eps_N2 = n2_params["A"] * T ** n2_params["n"] * math.exp(-n2_params["Ea"] / (R * T))
print(f"\nStep 1: eps_N2")
print(f"  = {n2_params['A']} * {T}^{n2_params['n']} * exp(-{n2_params['Ea']}/(1.987*300))")
print(f"  = {n2_params['A']} * {T ** n2_params['n']:.6e} * {math.exp(-n2_params['Ea'] / (R * T)):.6e}")
print(f"  eps_N2 = {eps_N2:.10f}")

# Step 2: sigmaX_M
X_N2 = composition["N2"]
X_Ar = composition["Ar"]
sigmaX_M = 1.0 - X_N2  # Only N2 is specified, so Ar goes into M
print(f"\nStep 2: sigmaX_M")
print(f"  sigmaX_M = 1.0 - X_N2 = 1.0 - {X_N2} = {sigmaX_M}")

# Step 3: eps_mix
eps_M = 1.0  # Always 1
eps_mix = X_N2 * eps_N2 + sigmaX_M * eps_M
print(f"\nStep 3: eps_mix")
print(f"  eps_mix = X_N2 * eps_N2 + sigmaX_M * eps_M")
print(f"  eps_mix = {X_N2} * {eps_N2:.10f} + {sigmaX_M} * {eps_M}")
print(f"  eps_mix = {eps_mix:.10f}")

# Step 4: Evaluate Plog at effective pressure
# For N2 (efficiency only, uses M's rate with eps2=eps_M=1)
P_eff_N2 = P * eps_mix / 1.0
print(f"\nStep 4a: P_eff for N2")
print(f"  P_eff_N2 = P * eps_mix / eps2")
print(f"  P_eff_N2 = {P} * {eps_mix:.10f} / 1.0 = {P_eff_N2:.10f} atm")

# Interpolate Plog at P_eff_N2
plog_data = {
    1.000e00: {"A": 7.72584e16, "n": -2.15195e00, "Ea": 2.13473e02},
    3.000e00: {"A": 2.11688e17, "n": -2.14062e00, "Ea": 2.79031e02},
}

P1, P2 = 1.0, 3.0
params1, params2 = plog_data[P1], plog_data[P2]

k1 = params1["A"] * T ** params1["n"] * math.exp(-params1["Ea"] / (R * T))
k2 = params2["A"] * T ** params2["n"] * math.exp(-params2["Ea"] / (R * T))

print(f"\nStep 4b: Interpolate Plog")
print(f"  At P={P1} atm: k1 = {k1:.10e}")
print(f"  At P={P2} atm: k2 = {k2:.10e}")

# Log-linear interpolation
log_k = math.log(k1) + (math.log(P_eff_N2) - math.log(P1)) / (math.log(P2) - math.log(P1)) * (
    math.log(k2) - math.log(k1)
)
k_M = math.exp(log_k)
print(f"  Interpolated k_M(P_eff={P_eff_N2:.10f}) = {k_M:.10e}")

# Step 5: Compute contributions
print(f"\nStep 5: Compute contributions")

# N2 contribution
contrib_N2 = k_M * eps_N2 * X_N2 / eps_mix
print(f"  N2 contribution:")
print(f"    = k_M * eps_N2 * X_N2 / eps_mix")
print(f"    = {k_M:.10e} * {eps_N2:.10f} * {X_N2} / {eps_mix:.10f}")
print(f"    = {contrib_N2:.10e}")

# Ar (M) contribution
contrib_Ar = k_M * eps_M * X_Ar / eps_mix
print(f"  Ar (M) contribution:")
print(f"    = k_M * eps_M * X_Ar / eps_mix")
print(f"    = {k_M:.10e} * {eps_M} * {X_Ar} / {eps_mix:.10f}")
print(f"    = {contrib_Ar:.10e}")

# Total
k_total = contrib_N2 + contrib_Ar
print(f"\nStep 6: Total rate constant")
print(f"  k_total = {contrib_N2:.10e} + {contrib_Ar:.10e}")
print(f"  k_total = {k_total:.15e}")
print(f"  k_total = {k_total:.10f}")

print(f"\n" + "=" * 60)
print(f"CANTERA: 329171038824.976074")
print(f"OURS:    {k_total:.6f}")
print(f"DIFF:    {abs(329171038824.976074 - k_total):.6f}")
print("=" * 60)
````
