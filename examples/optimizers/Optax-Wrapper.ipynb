```python
import jax
import jax.numpy as jnp
from diffPLOG2TROE.optimizers.optax_wrapper import OptaxWrapper

jax.config.update("jax_enable_x64", True)
```

## Adam basic

```python
def loss_function(params):
    return jnp.sum((params - 3.0) ** 2)

# Initialize parameters
params = jnp.array([1.0, 2.0, -1.0, 0.0])
active_indices = jnp.array([0, 1, 2, 3])  # Optimize all parameters

# Configure Adam optimizer
adam_options = {
    "algorithm": "adam",
    "max_steps": 1000,
    "patience": 50,
    "ftol": 1e-6,
    "steps_file": 10,
    "learning_rate": 0.1
}

# Create optimizer and run optimization
optimizer = OptaxWrapper(adam_options)
result = optimizer.optimize(loss_function, params, active_indices)

print(f"Final loss: {result['loss']:.6e}")
print(f"Optimized parameters: {result['params']}")
print(f"Number of iterations: {result['iterations']}")
```

## Adam with learning rate scheduler

```python
# Configure Adam with cosine learning rate schedule
adam_schedule_options = {
    "algorithm": "adam",
    "max_steps": 5000,
    "patience": 100,
    "ftol": 1e-7,
    "steps_file": 50,
    "learning_rate": 0.01,
    "clip_norm": 1.0,
    "use_scheduler": True,
    "schedule_type": "cosine",
    "warmup_steps": 500,
    "decay_steps": 4500,
    "peak_value": 0.1,
    "end_value": 1e-5,
    "init_value": 1e-4
}

# Create optimizer and run optimization
optimizer = OptaxWrapper(adam_schedule_options)
result = optimizer.optimize(loss_function, params, active_indices)

print(f"Final loss: {result['loss']:.6e}")
print(f"Optimized parameters: {result['params']}")
print(f"Number of iterations: {result['iterations']}")
```
